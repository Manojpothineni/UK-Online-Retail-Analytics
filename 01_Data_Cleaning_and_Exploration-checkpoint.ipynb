```python
# Create a copy for cleaning
df_clean = df.copy()
initial_rows = len(df_clean)

print(f"Initial rows: {initial_rows:,}\n")

# Step 1: Convert InvoiceDate to datetime
print("Step 1: Converting InvoiceDate to datetime...")
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])
print(" InvoiceDate converted to datetime\n")

# Step 2: Remove cancelled orders
print("Step 2: Removing cancelled orders (InvoiceNo starting with 'C')...")
df_clean = df_clean[~df_clean['InvoiceNo'].astype(str).str.startswith('C')]
print(f" Removed {initial_rows - len(df_clean):,} cancelled orders")
print(f" Remaining rows: {len(df_clean):,}\n")

# Step 3: Remove rows with bad UnitPrice
print("Step 3: Removing zero/negative prices...")
before = len(df_clean)
df_clean = df_clean[df_clean['UnitPrice'] > 0]
print(f" Removed {before - len(df_clean):,} rows with bad prices")
print(f" Remaining rows: {len(df_clean):,}\n")

# Step 4: Remove rows with zero quantity
print("Step 4: Removing zero quantities...")
before = len(df_clean)
df_clean = df_clean[df_clean['Quantity'] > 0]
print(f" Removed {before - len(df_clean):,} rows with zero quantity")
print(f" Remaining rows: {len(df_clean):,}\n")

# Step 5: Remove duplicates
print("Step 5: Removing duplicate rows...")
before = len(df_clean)
df_clean = df_clean.drop_duplicates()
print(f" Removed {before - len(df_clean):,} duplicate rows")
print(f" Remaining rows: {len(df_clean):,}\n")

# Step 6: Remove rows with missing Description
print("Step 6: Removing rows with missing Description...")
before = len(df_clean)
df_clean = df_clean[df_clean['Description'].notna()]
print(f" Removed {before - len(df_clean):,} rows with missing description")
print(f" Remaining rows: {len(df_clean):,}\n")

# Step 7: Clean Description (remove extra spaces)
print("Step 7: Cleaning Description field...")
df_clean['Description'] = df_clean['Description'].str.strip().str.upper()
print(" Description field cleaned\n")

print(f"Original rows:  {initial_rows:>5,}")
print(f"Cleaned rows:   {len(df_clean):>5,}")
print(f"Rows removed:   {initial_rows - len(df_clean):>5,}")
```

```python
# 1. TotalPrice (Revenue)
df_clean['TotalPrice'] = df_clean['Quantity'] * df_clean['UnitPrice']
# 2. Year
df_clean['Year'] = df_clean['InvoiceDate'].dt.year

# 3. Month
df_clean['Month'] = df_clean['InvoiceDate'].dt.month


# 4. MonthName
df_clean['MonthName'] = df_clean['InvoiceDate'].dt.month_name()


# 5. Day
df_clean['Day'] = df_clean['InvoiceDate'].dt.day


# 6. DayOfWeek
df_clean['DayOfWeek'] = df_clean['InvoiceDate'].dt.day_name()

# 7. Hour
df_clean['Hour'] = df_clean['InvoiceDate'].dt.hour

# 8. Quarter
df_clean['Quarter'] = df_clean['InvoiceDate'].dt.quarter

# 9. IsWeekend
df_clean['IsWeekend'] = df_clean['DayOfWeek'].isin(['Saturday', 'Sunday'])

# 10. Date (without time)
df_clean['Date'] = df_clean['InvoiceDate'].dt.date

print(f"\n New dataset shape: {df_clean.shape[0]:,} rows × {df_clean.shape[1]} columns")

# Display sample with new columns
df_clean.head()
```

```python
# Monthly revenue analysis
monthly_revenue = df_clean.groupby('MonthName')['TotalPrice'].sum().sort_values(ascending=False)

print(monthly_revenue)

# Plot
plt.figure(figsize=(12, 6))
monthly_revenue.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Total Revenue by Month', fontsize=16, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Revenue (£)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\n🏆 Highest Revenue Month: {monthly_revenue.idxmax()} (£{monthly_revenue.max():,.2f})")
print(f"📉 Lowest Revenue Month: {monthly_revenue.idxmin()} (£{monthly_revenue.min():,.2f})")
```

```python
# Top 10 products by revenue
top_products = df_clean.groupby('Description').agg({
    'TotalPrice': 'sum',
    'Quantity': 'sum'
}).sort_values('TotalPrice', ascending=False).head(10)


print("TOP 10 PRODUCTS BY REVENUE")

print(top_products)

# Plot
plt.figure(figsize=(12, 6))
top_products['TotalPrice'].plot(kind='barh', color='coral', edgecolor='black')
plt.title('Top 10 Products by Revenue', fontsize=16, fontweight='bold')
plt.xlabel('Revenue (£)', fontsize=12)
plt.ylabel('Product', fontsize=12)
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Revenue by country
country_revenue = df_clean.groupby('Country')['TotalPrice'].sum().sort_values(ascending=False)

print("TOP 10 COUNTRIES BY REVENUE")
print(country_revenue.head(10))

# Calculate UK percentage
uk_revenue = country_revenue.get('United Kingdom', 0)
total_revenue = country_revenue.sum()
uk_pct = (uk_revenue / total_revenue) * 100

print(f"\n🇬🇧 UK Revenue: £{uk_revenue:,.2f} ({uk_pct:.2f}% of total)")
print(f"🌍 International Revenue: £{total_revenue - uk_revenue:,.2f} ({100-uk_pct:.2f}% of total)")

# Plot
plt.figure(figsize=(12, 6))
country_revenue.head(10).plot(kind='bar', color='green', edgecolor='black')
plt.title('Top 10 Countries by Revenue', fontsize=16, fontweight='bold')
plt.xlabel('Country', fontsize=12)
plt.ylabel('Revenue (£)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
